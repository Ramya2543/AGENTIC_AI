# Day 2 – Prompts, Memory & Python for AI Agents

## Topic 1: Messages and Roles

Every AI conversation is represented using messages.

Each message contains:
- **system:** Instructions given to the AI.
- **user:** The message sent by the user.
- **assistant:** The AI's previous response.

The AI reads the conversation history and generates its next response.

We are using the Sweet Crumbs Home Bakery chatbot as our example.

```python
messages = [
    {"role": "system", "content": "You are a friendly bakery assistant."},
    {"role": "user", "content": "Do you have eggless cakes?"},
    {"role": "assistant", "content": "Yes, all our cakes are eggless!"}
]

for message in messages:
    print(message["role"], ":", message["content"])
```

## Topic 2: System Prompt

A system prompt provides instructions to an AI about how it should behave.

A good system prompt can contain:

- **ROLE:** Defines who the AI is.
- **FACTS:** Provides information the AI should know.
- **RULES:** Defines what the AI must or must not do.
- **FORMAT:** Defines how the AI should respond.

We are creating a system prompt for Sweet Crumbs Home Bakery.

```python
MENU = {
    "plum cake": 500,
    "chocolate cake": 650,
    "cupcake": 80
}

SYSTEM_PROMPT = f"""
ROLE: You are a friendly assistant for Sweet Crumbs Home Bakery.

FACTS:
- Plum cake costs ₹500.
- Chocolate cake costs ₹650.
- Cupcake costs ₹80.
- Orders need 24 hours notice.
- All cakes have eggless versions.

RULES:
- Never promise a delivery time.
- If you don't know something, ask the owner.
- Be polite and helpful.

FORMAT:
- Keep answers short.
- Use friendly language.
- Reply in the customer's language.
"""

print(SYSTEM_PROMPT)
```

```python
history = [
    {"role": "system", "content": "You are a friendly bakery assistant."}
]

MAX_TURNS = 2

# Add customer messages
history.append({"role": "user", "content": "My name is Ramya."})
history.append({"role": "assistant", "content": "Nice to meet you, Ramya!"})

history.append({"role": "user", "content": "I want a chocolate cake."})
history.append({"role": "assistant", "content": "Sure! Chocolate cake costs Rs 650."})

history.append({"role": "user", "content": "How much does it cost?"})

# Keep the system message and recent messages
recent_history = [history[0]] + history[1:][-MAX_TURNS:]

print("Complete History:")
print(history)

print("\nTrimmed History:")
print(recent_history)
```

```python
MENU = {
    "plum cake": 500,
    "chocolate cake": 650,
    "cupcake": 80
}

# Accessing a dictionary value
print("Plum cake price:", MENU["plum cake"])

# Safe lookup using get()
print("Donut price:", MENU.get("donut", "Not on menu"))

# Creating an order dictionary
order = {
    "name": "Ravi",
    "item": "plum cake",
    "qty": 2,
    "eggless": True
}

# Calculating total price
total = order["qty"] * MENU[order["item"]]
print("Total price:", total)

# Dictionary to JSON string
import json

text = json.dumps(order)
print("\nJSON string:")
print(text)

# JSON string back to dictionary
back = json.loads(text)
print("\nPython dictionary:")
print(back)
```

```python
MENU = {
    "plum cake": 500,
    "chocolate cake": 650,
    "cupcake": 80
}

def get_price(item: str, qty: int = 1) -> int:
    """Return the total price in rupees for qty of an item."""
    return MENU[item.lower().strip()] * qty

print(get_price("plum cake"))
print(get_price("chocolate cake", 2))
print(get_price("cupcake", 5))
```

```python
def my_decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
```
